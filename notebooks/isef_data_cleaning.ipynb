```python
def clean_school(val):
    if pd.isna(val):
        return np.nan
    try:
        val = str(val)
        val = val.replace("{","").replace("}","").strip()
        return val
    except:
        return np.nan
    
```

```python
df["schools"] = df["schools"].apply(clean_school)
```

```python
def clean_nan(val):
    if pd.isna(val):
        return np.nan
    val = str(val)
    val = val.replace("[","").replace("]","").strip()
    if "nan" in val:
        return np.nan
    return val
```

```python
df["awards"] = df["awards"].apply(clean_nan)
df.head()
```

```python
df.info()
```

```python
def clean_schools(val):
    if pd.isna(val):
        return np.nan
    if ("set" in val) and (isinstance(val,str)):
        return np.nan
```

```python
df["schools"] = df["schools"].apply(clean_schools)
df.head()
```
